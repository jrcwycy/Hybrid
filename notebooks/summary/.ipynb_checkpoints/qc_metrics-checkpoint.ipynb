## Hybrid

```python
hdata = adata[adata.obs['dataset'] == 'Hybrid', :].copy()

# remove genes not expressed 
sc.pp.filter_genes(hdata, min_cells=1)
hdata
```

```python
### --- Handle HYB cells with NaN feature barcode predictions ---
init_cells_hybrid = hdata[hdata.obs['dataset'] == 'Hybrid'].n_obs

# remove HYB cells with NaN feature barcode predictions
hybrid_to_remove = (hdata.obs['dataset'] == 'Hybrid') & (hdata.obs['assigned_condition'].isna())
cells_to_keep = ~(hybrid_to_remove)
hdata = hdata[cells_to_keep].copy()

final_cells_hybrid = hdata[hdata.obs['dataset'] == 'Hybrid'].n_obs
removed_hybrid_cells = init_cells_hybrid - final_cells_hybrid

print(f"Removed {removed_hybrid_cells} 'Hybrid' cells with unassigned feature barcodes")
print(f"Final shape: {hdata.shape}\n")
```

```python
# cell counts
hdata.obs['assigned_condition'].value_counts()
```

```python
## gene counts per condition
nonzero_matrix = hdata.X > 0

# Create a Series that maps each observation to its dataset
datasets = hdata.obs['assigned_condition']

# Convert to DataFrame for grouping
nonzero_df = pd.DataFrame(nonzero_matrix.toarray() if hasattr(nonzero_matrix, "toarray") else nonzero_matrix,
                          index=hdata.obs_names,
                          columns=hdata.var_names)
nonzero_df['assigned_condition'] = datasets.values

# Group by dataset and count number of genes (columns) with any non-zero value
genes_detected = nonzero_df.groupby('assigned_condition', observed=True).apply(lambda x: (x.drop(columns='assigned_condition').sum(axis=0) > 0).sum())

print(f"Number of non-zero genes:\n {genes_detected}\n")
```

# Processed data

## Hybrid

```python
# mitochondrial genes
hdata.var["mt"] = hdata.var_names.str.startswith("MT-")
# ribosomal genes
hdata.var["ribo"] = hdata.var_names.str.startswith(("RPS", "RPL"))
# hemoglobin genes
hdata.var["hb"] = hdata.var_names.str.contains("^HB[^(P)]")

sc.pp.calculate_qc_metrics(hdata, qc_vars=["mt", "ribo", "hb"], inplace=True, log1p=True)

plt.rcdefaults()
sc.pl.violin(
    hdata,
    ["n_genes_by_counts", "total_counts", "pct_counts_mt"],
    jitter=0.4,
    multi_panel=True,
    groupby='assigned_condition',
)
```

```python
def is_outlier(adata, metric: str, nmads: int):
    M = adata.obs[metric]

    if len(M) < 2 or M.isnull().all():
        return pd.Series(False, index=adata.obs.index)

    median_val = np.median(M)
    mad_val = median_abs_deviation(M)
    print(f'{metric}:\tMedian val: {median_val}, MAD val: {mad_val}')

    if mad_val == 0:
        return pd.Series(False, index=adata.obs.index)

    outliers = (M < median_val - nmads * mad_val) | \
               (M > median_val + nmads * mad_val)

    return outliers



### determine outliers
hdata.obs["outlier"] = (
    is_outlier(hdata, "log1p_total_counts", 5)
    | is_outlier(hdata, "log1p_n_genes_by_counts", 5)
)

print("Number of outlier cells based on counts:")
print(hdata.obs['outlier'].sum())


hdata.obs["mt_outlier"] = (
    is_outlier(hdata, "pct_counts_mt", 5)
    | (hdata.obs["pct_counts_mt"] > 20)
)

print("\nNumber of outlier cells based on pct mito:")
print(hdata.obs['mt_outlier'].sum())


init_cells = hdata.n_obs
print(f"Total number of cells: {init_cells}")

print("\nFiltering...")
hdata = hdata[(~hdata.obs.outlier) & (~hdata.obs.mt_outlier)].copy()

final_cells = hdata.n_obs

print(f"\nRemoved {init_cells - final_cells} total cells.")
print(f"\nTotal cells after filtering of low quality cells: {final_cells}")
```

```python
# cell counts
hdata.obs['assigned_condition'].value_counts()
```

```python
# number of genes with at least 10 counts per condition
counts_df = pd.DataFrame(
    hdata.X.toarray() if hasattr(hdata.X, "toarray") else hdata.X,
    index=hdata.obs_names,
    columns=hdata.var_names
)
counts_df["assigned_condition"] = hdata.obs["assigned_condition"].values

genes_detected = (
    counts_df.groupby("assigned_condition", observed=True)
    .apply(lambda x: (x.drop(columns="assigned_condition").sum(axis=0) >= 10).sum())
)

print(genes_detected)
```

```python
results = []

for condition in hdata.obs["assigned_condition"].cat.categories:
    # Subset to one condition
    ad = hdata[hdata.obs["assigned_condition"] == condition]

    # Total counts per gene across all cells
    gene_totals = np.asarray(ad.X.sum(axis=0)).ravel()

    # Keep genes with >=10 total counts
    keep_genes = gene_totals >= 10

    # Subset to retained genes
    X = ad[:, keep_genes].X

    # Genes detected (>0 counts) per cell
    genes_per_cell = np.asarray((X > 0).sum(axis=1)).ravel()

    # Total counts per cell
    counts_per_cell = np.asarray(X.sum(axis=1)).ravel()

    results.append({
        "assigned_condition": condition,
        "n_genes_>=10_counts": keep_genes.sum(),
        "mean_genes_per_cell": genes_per_cell.mean(),
        "median_genes_per_cell": np.median(genes_per_cell),
        "mean_counts_per_cell": counts_per_cell.mean(),
        "median_counts_per_cell": np.median(counts_per_cell),
    })

results_df = pd.DataFrame(results)
display(results_df)
```
