# Scree plots

```python
gene_sets = {
    'TFs': tf_list,
    'cell_cycle': cell_cycle_genes,
    'fibroblast': fib_genes,
    'muscle': muscle_genes,
    'hvg_2000': hvg_top2000,
    'hvg_500': hvg_top500,
}

cond_order = ['Control', 'mmMYOD1', 'siPRRX1/mmMYOD1', 'siPRRX1']
colors = {'Control': '#D3D4D8', 'mmMYOD1': '#C52E19', 'siPRRX1/mmMYOD1': '#FBA72A', 'siPRRX1': '#5785C1'}

# results[gene_set][cond] = {...}
results = {}


n_comps = 50
n_top_pcs = 5


for gene_set, gene_list in gene_sets.items():
    
    gene_list = [g for g in gene_list if g in adata.var_names]
    if len(gene_list) < 2:
        print(f"Skipping {gene_set}: only {len(gene_list)} genes found in adata.var_names")
        continue
    
    tmp = adata[:, gene_list].copy()
    results[gene_set] = {}
    
    results[gene_set]['n_genes'] = len(gene_list)
    
    for cond in cond_order:
        bdata = tmp[tmp.obs['pooled_condition'] == cond, :].copy()
        if bdata.n_obs < 3:
            print(f"Skipping {gene_set} / {cond}: too few cells ({bdata.n_obs})")
            continue
        # print(f"Data shape: {bdata.shape}")

        bdata.X = bdata.layers['log_norm'].copy()
        sc.pp.scale(bdata) # zero mean, unit variance
        sc.pp.pca(bdata, n_comps=n_comps, mask_var=None)  # computes SVD of the scaled matrix: U∑V^T

         # Extract singular values (stored as square root of variance)
        n_cells = bdata.n_obs
        
        singular_values_squared = bdata.uns['pca']['variance'] * (n_cells - 1)
        singular_values = np.sqrt(singular_values_squared)

        # Extract loadings (top contributing genes)
        loadings = bdata.varm['PCs'][:, :n_top_pcs] # shape: (n_genes, n_top_pcs)
        loadings_df = pd.DataFrame(
            loadings,
            index=bdata.var_names,
            columns=[f'PC{i+1}' for i in range(n_top_pcs)]
        )

        results[gene_set][cond] = {
            'singular_values': singular_values,
            'normalized_sv': singular_values / singular_values.sum(),
            'singular_values_squared': singular_values_squared,
            'normalized_sv_squared': singular_values_squared / singular_values_squared.sum(),
            'n_pcs': len(singular_values),
            'loadings': loadings_df,
        }
```

```python
# --------- Plotting ---------

max_pc_to_plot = 12

gene_set_names = list(results.keys())
n_sets = len(gene_set_names)


ncols = 2
nrows = math.ceil(n_sets / ncols)

fig, axes = plt.subplots(nrows, ncols, figsize=(4.3 * ncols, 3.9 * nrows))#, sharey=True)
axes = np.atleast_1d(axes).ravel()


for i, gene_set in enumerate(gene_set_names):
    ax = axes[i]
    
    for cond in cond_order:
        if cond not in results[gene_set]:
            continue
        
        y = results[gene_set][cond]['normalized_sv_squared']
        n_pcs = min(len(y), max_pc_to_plot)
        
        ax.plot(
            range(1, n_pcs + 1), y[:n_pcs],
            marker='o', markersize=3.5, linewidth=1,
            color=colors[cond], label=cond
        )
        
    ax.set_title(f'{gene_set} genes ($N_g$ = {results[gene_set]['n_genes']})')
    ax.set_xlabel('Component')
    ax.set_ylabel(r'$σ_i^2 / \sum σ_i^2$')
    ax.set_xlim(0.5, max_pc_to_plot + 0.5)

    if i == 0:
        ax.legend(loc='upper right', fontsize=9, frameon=True)
    
for j in range(n_sets, len(axes)):
    axes[j].axis('off')

    
plt.tight_layout()
plt.show()
```

# Correlation

## determine which genes are expressed

```python
# Determine which genes are "expressed"
aggdata = sc.get.aggregate(
    adata,
    by='pooled_condition',
    func='sum',
    layer='raw_counts',
)

aggdata.X = aggdata.layers['sum'].copy()

sc.pp.normalize_total(aggdata, target_sum=1e4)
aggdata.layers['CP10K'] = aggdata.X.copy()

sc.pp.log1p(aggdata)
aggdata.layers['log_norm'] = aggdata.X.copy()

print(aggdata)

aggdata.X = aggdata.layers['log_norm'].copy()

X = aggdata.X if isinstance(aggdata.X, np.ndarray) else aggdata.X.toarray()

df = pd.DataFrame(X, index=aggdata.obs_names, columns=aggdata.var_names).T

df = df.reset_index()

print(df.shape)
df.head()
```

```python
mdf = df[df['gene_name'].isin(fib_genes)].copy()

# ---------- Add % expressing per condition columns ----------
def pct_expressing_by_group(adata, group_key, genes, layer="raw_counts", threshold=0):
    # subset to genes of interest
    genes = [g for g in genes if g in adata.var_names]
    A = adata[:, genes].layers[layer]
    if not sp.issparse(A):
        A = sp.csr_matrix(A)

    groups = adata.obs[group_key].astype(str).values
    out = pd.DataFrame(index=genes)

    for g in np.unique(groups):
        idx = np.where(groups == g)[0]
        sub = A[idx, :]
        pct = (sub > threshold).mean(axis=0)  # fraction expressing
        pct = np.asarray(pct).ravel()
        out[f"pct_expr_{g}"] = pct

    out.index.name = "gene_name"
    return out.reset_index()

# compute for MYOD network genes
pct_df = pct_expressing_by_group(
    adata, group_key="pooled_condition", genes=fib_genes, layer="raw_counts", threshold=0
)

mdf = mdf.merge(pct_df, on='gene_name', how='left')
print(mdf.shape)
mdf.head()
```

```python
# ---------- Set expr to 0 for genes expressed in less than 10% of cells per group ----------
conditions = ['Control', 'mmMYOD1', 'siPRRX1', 'siPRRX1/mmMYOD1']
pct_prefix = 'pct_expr_'
min_pct = 0.10 # 10%

for cond in conditions:
    pct_col = f"{pct_prefix}{cond}"
    expr_col = cond
    mdf.loc[mdf[pct_col] < min_pct, expr_col] = 0
    
display(mdf.head())

print(f"\nN genes expressed in >= 10% of cells for...\n")

total_genes = mdf.shape[0]

for col in conditions:
    nonzero_genes = (mdf[col] != 0).sum()
    print(f"{col}: {nonzero_genes} genes\n% of total = {nonzero_genes/total_genes*100:.2f}%\n")
```

```python
ctrl_genes = mdf[mdf['Control'] != 0]['gene_name'].unique()

mmy_genes = mdf[mdf['mmMYOD1'] != 0]['gene_name'].unique()
prx_genes = mdf[mdf['siPRRX1'] != 0]['gene_name'].unique()
hyb_genes = mdf[mdf['siPRRX1/mmMYOD1'] != 0]['gene_name'].unique()


# ------------ final gene list ------------

# shared by Control and mmMYOD1
gene_list = list(set(ctrl_genes) & set(mmy_genes))
print(len(gene_list))
```

## Compute correlation

```python
tmp = adata[:, gene_list].copy()
print(tmp.shape)

corr_method = 'spearman'

corr_matrices = {}

conditions = ['Control', 'mmMYOD1']

for cond in conditions:
    
    tmp1 = tmp[tmp.obs['pooled_condition'] == cond, :].copy()
    print(tmp1.shape)
    tmp1.X = tmp1.layers['log_norm'].copy()
    
    X = tmp1.X if isinstance(tmp1.X, np.ndarray) else tmp1.X.toarray()

    X_df = pd.DataFrame(
        X,
        index=tmp1.obs_names,
        columns=tmp1.var_names,
    )

    RM = X_df.corr(method=corr_method)
    print(RM.shape)
    corr_matrices[cond] = RM
```

```python
display(corr_matrices['Control'].head())

display(corr_matrices['mmMYOD1'].head())
```

## Fisher Z-transform

```python
c1, c2 = 'Control', 'mmMYOD1'

RM1 = corr_matrices[c1].copy()
RM2 = corr_matrices[c2].copy()

genes = RM1.index.intersection(RM2.index)
RM1 = RM1.loc[genes, genes]
RM2 = RM2.loc[genes, genes]


eps = 1e-7
R1 = RM1.to_numpy(dtype=float)
R2 = RM2.to_numpy(dtype=float)

R1 = np.clip(R1, -1 + eps, 1 - eps)
R2 = np.clip(R2, -1 + eps, 1 - eps)

Z1 = 0.5 * np.log((1.0 + R1) / (1.0 - R1))
Z2 = 0.5 * np.log((1.0 + R2) / (1.0 - R2))

DZ = Z2 - Z1

ADZ = np.abs(DZ)

# ------------ genes with biggest overall change (aggregate across pairs) ------------
gene_score = np.nanmean(ADZ, axis=1)  # mean abs z-change for each gene vs all others
gene_rank = pd.Series(gene_score, index=genes).sort_values(ascending=False)
print(gene_rank.head(20))
```

```python
# ------------ gene-gene pairs (edges) with biggest change ------------
iu = np.triu_indices_from(ADZ, k=1)
edge_df = pd.DataFrame({
    "gene1": genes[iu[0]],
    "gene2": genes[iu[1]],
    "dz": DZ[iu],           # signed change
    "abs_dz": ADZ[iu],      # magnitude
}).sort_values("abs_dz", ascending=False)

print(edge_df.head(20))
```
