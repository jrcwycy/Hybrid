```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.sparse as sp
import time
import anndata as an
import scanpy as sc
from sklearn.metrics import confusion_matrix, classification_report
from sklearn.metrics import balanced_accuracy_score, cohen_kappa_score, matthews_corrcoef
```

# Load data

```python
%%time
path = "/nfs/turbo/umms-indikar/shared/projects/hybrid_reprogramming/anndata/processed_all_groups_v2.h5ad"
adata = sc.read_h5ad(path)
sc.logging.print_memory_usage()
adata
```

```python
adata.obs[adata.obs['pooled_condition'] == 'Control']['assigned_condition'].value_counts(normalize=True)
```

# Confusion matrix

```python
# sc.pp.scale was run on log-norm before calling cc phase
df = adata.obs[['assigned_condition', 'dataset', 'S_score', 'G2M_score', 'phase']].copy()

df = df[df['dataset'] == 'Control']
print(f'Total cells: {df.shape[0]}')

# drop rows with missing labels
df = df.dropna(subset=['assigned_condition', 'phase'])

print(f'After dropping missing labels: {df.shape[0]}')
df.head()
```

```python
df['assigned_condition'] = df['assigned_condition'].astype(str)
df['phase'] = df['phase'].astype(str)

print(df['assigned_condition'].value_counts(normalize=True))
print(df['phase'].value_counts(normalize=True))
```

```python
labels = ['G1', 'S', 'G2M'] #sorted(set(df["phase"]) | set(df["assigned_condition"]))

# Compute confusion matrix
cm = confusion_matrix(
    df["assigned_condition"], # y_true
    df["phase"], # y_pred
    labels=labels
)

# Convert to DataFrame for nicer plotting
cdf = pd.DataFrame(
    cm,
    index=labels,
    columns=labels
)

print(cdf.shape)
display(cdf)


plt.figure(figsize=(4, 3.5), dpi=150)
sns.heatmap(
    cdf,
    annot=True,
    fmt="d",
    cmap="Blues",
    cbar=True
)

plt.ylabel("Barcoded phase")
plt.xlabel("Predicted phase")
plt.title("Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
cm_norm = confusion_matrix(
    df["assigned_condition"],
    df["phase"],
    labels=labels,
    normalize="true"
)

cndf = pd.DataFrame(
    cm_norm,
    index=labels,
    columns=labels
)

display(cndf)

plt.figure(figsize=(4, 3.5), dpi=150)
sns.heatmap(
    cndf,
    annot=True,
    fmt=".2f",
    cmap="Blues",
    cbar=True
)

plt.ylabel("Barcoded phase")
plt.xlabel("Predicted phase")
plt.title("Normalized Confusion Matrix (w/ scaling)")
plt.tight_layout()
plt.show()
```

```python
sns.violinplot(
    data=df,
    x='assigned_condition',
    y='S_score',
    order=['G1', 'S', 'G2M']
)
plt.axhline(y=0, ls='--', lw=1)

plt.xlabel("Barcoded phase")
plt.ylabel("S score")

plt.show()

sns.violinplot(
    data=df,
    x='assigned_condition',
    y='G2M_score',
    order=['G1', 'S', 'G2M']
)
plt.axhline(y=0, ls='--', lw=1)

plt.xlabel("Barcoded phase")
plt.ylabel("G2M score")

plt.show()
```

```python
df['assigned_condition'] = df['assigned_condition'].astype(str)

sns.scatterplot(
    data=df,
    x='S_score',
    y='G2M_score',
    hue='assigned_condition',
    alpha=0.6,
    s=7,
    ec='none',
)
plt.axhline(0, color='gray', linestyle='--')
plt.axvline(0, color='gray', linestyle='--')
plt.show()
```

```python
labels = ['G1', 'S', 'G2M']

y_true = df['assigned_condition']
y_pred = df['phase']  # or your saved predicted phase column

print(confusion_matrix(y_true, y_pred, labels=labels))
print(classification_report(y_true, y_pred, labels=labels))

print("Balanced accuracy:", balanced_accuracy_score(y_true, y_pred))
print("Cohen kappa:", cohen_kappa_score(y_true, y_pred, labels=labels))
print("MCC:", matthews_corrcoef(y_true, y_pred))
```

```python
cdf = pd.read_csv("../../resources/GO_2025_cc_genes.csv")

print(cdf.shape)
cdf.head()
```

```python
S_cols = [
    'G1S_transition',
    'DNA_replication_S_phase',
]

S_genes = cdf.loc[cdf[S_cols].any(axis=1), 'gene'].tolist()

print(len(S_genes))


g2m_cols = [
    'G2M_transition',
    'metaphase_anaphase_transition',
]

G2M_genes = cdf.loc[cdf[g2m_cols].any(axis=1), 'gene'].tolist()

print(len(G2M_genes))
```

```python
print(len(list(set(G2M_genes) & set(S_genes))))

G2M_genes1 = list(set(G2M_genes) - set(S_genes))
print(len(G2M_genes1))
```

```python
S_genes1 = list(set(S_genes) - set(G2M_genes))
print(len(S_genes1))
```

```python
# with scaling

adata.X = adata.layers['log_norm'].copy()

sc.pp.scale(adata)

sc.tl.score_genes_cell_cycle(
    adata,
    s_genes=S_genes1,
    g2m_genes=G2M_genes1,
)

adata.obs[['phase']].head()
```
