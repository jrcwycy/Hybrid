Compute nearest-neighbor displacement distribution between consecutive frames
* Goal: Emperically determine what max_search_radius should be

# Load regionprops tables

```python
%%time
dpath = "/scratch/indikar_root/indikar1/shared_data/hybrid_imaging/regionprops/"
file_list = glob.glob(f"{dpath}*_*_*.csv") # exclude the test image

df_list = []

for fpath in file_list:
    basename = os.path.basename(fpath).replace(".csv", "")
    stage, condition, scene, _ = basename.split("_")
    
    df = pd.read_csv(fpath)
    df['basename'] = basename
    df['stage'] = int(stage)
    df['condition'] = condition
    df['scene'] = scene
    
    print(f"\t{basename} {df.shape=}")

    df_list.append(df)


df = pd.concat(df_list)
print(f"{df.shape=}")

df['cell_id'] = df['basename'] + "_label" + df['label'].astype(str)

# shift stage 2 times to follow after stage 1
max_time_stage1 = df.loc[df["stage"] == 1, "time"].max()
df["time_shifted"] = df["time"] + (
    (df["stage"] == 2) * (max_time_stage1 + 1)
)

df.head()
```

```python
print(df.columns)
```

```python
columns_to_keep = [
    'centroid_x_px', 'centroid_y_px',
    'cell_id', 'basename', 'time', 
    'area', "major_axis_length", 
    "minor_axis_length", 
    "orientation", 
    "solidity",
    "eccentricity",
]

sdf = df[columns_to_keep].copy()

sdf['well'] = sdf['basename'].str.replace("_props", "", regex=False)

sdf = sdf.rename(columns={'cell_id': 'ID', 'centroid_x_px': 'x', 'centroid_y_px': 'y', 'time': 'frame'})

sdf = sdf.drop(columns='basename')

print(sdf.shape)
sdf.head()
```

```python
def diagnose_displacements_df(
    df,
    well_col: str = "well",
    frame_col: str = "frame",
    x_col: str = "x",
    y_col: str = "y",
    frame_interval_min: float = 20.0,
):
    """
    Same diagnostic as diagnose_displacements(), but runs directly off a
    dataframe with columns for well/frame/x/y, and reports per-well AND
    pooled statistics (per-well matters -- density/motility can differ
    a lot between wells/conditions).
 
    Returns a dict: {well_id: np.ndarray of NN displacements}
    """
    per_well_dists = {}
 
    for well_id, well_df in df.groupby(well_col):
        by_frame = {
            f: g[[x_col, y_col]].to_numpy()
            for f, g in well_df.groupby(frame_col)
        }
        frames = sorted(by_frame.keys())
 
        if frames != list(range(frames[0], frames[-1] + 1)):
            print(f"[{well_id}] WARNING: frame indices not contiguous "
                  f"({len(frames)} unique frames spanning "
                  f"{frames[0]}-{frames[-1]}). Check frame numbering for this well.")
 
        nn_dists = []
        for f0, f1 in zip(frames[:-1], frames[1:]):
            pts0, pts1 = by_frame[f0], by_frame[f1]
            if len(pts0) == 0 or len(pts1) == 0:
                continue
            d = np.linalg.norm(pts0[:, None, :] - pts1[None, :, :], axis=-1)
            nn_dists.extend(d.min(axis=1).tolist())
 
        nn_dists = np.array(nn_dists)
        per_well_dists[well_id] = nn_dists
 
        if len(nn_dists) == 0:
            print(f"[{well_id}] no valid consecutive-frame pairs, skipping.")
            continue
 
        pct = np.percentile(nn_dists, [5, 25, 50, 75, 90, 95, 99])
        print(f"[{well_id}] n_frames={len(frames)}  "
              f"median={pct[2]:.1f}  p90={pct[4]:.1f}  p95={pct[5]:.1f}  p99={pct[6]:.1f}")
 
    all_dists = np.concatenate([d for d in per_well_dists.values() if len(d)])
    pct = np.percentile(all_dists, [5, 25, 50, 75, 90, 95, 99])
    print()
    print(f"POOLED across all wells ({frame_interval_min} min/frame):")
    print(f"  p5={pct[0]:.1f}  p25={pct[1]:.1f}  median={pct[2]:.1f}  "
          f"p75={pct[3]:.1f}  p90={pct[4]:.1f}  p95={pct[5]:.1f}  p99={pct[6]:.1f}")
    print()
    print(f"Suggested max_search_radius starting point: ~{pct[5]:.0f}-{pct[6]:.0f} "
          f"(p95-p99). Compare this to your current config's max_search_radius, "
          f"and double check your x/y units (pixels vs microns) match what the "
          f"config assumes.")
 
    return per_well_dists
```

```python
def load_objects_for_well(npy_path: str) -> "list[btypes.PyTrackObject]":
    """
    Convert your per-well .npy segmentation/metadata into a list of
    btrack PyTrackObject instances.
 
    Each object needs at minimum: ID, x, y, t (frame index), and ideally
    a 'label' / area / other features you want the tracker to use.
 
    Replace the body below with logic matching your actual format.
    """
    data = np.load(npy_path, allow_pickle=True).item()  # <-- adjust
 
    objects = []
    obj_id = 0
    for frame_idx, frame_data in enumerate(data["frames"]):  # <-- adjust
        for cell in frame_data["cells"]:                     # <-- adjust
            obj = btypes.PyTrackObject.from_dict(
                {
                    "ID": obj_id,
                    "t": frame_idx,
                    "x": cell["centroid_x"],   # <-- adjust key names
                    "y": cell["centroid_y"],
                    "z": 0.0,
                    "label": 5,  # 5 = generic object class in btrack
                }
            )
            objects.append(obj)
            obj_id += 1
    return objects
```
